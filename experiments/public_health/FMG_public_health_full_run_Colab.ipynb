# FMG public-health extension: full Colab run

Runs the pre-specified CDC BRFSS 2024 and UCI Diabetes 130-US Hospitals audit. It uses three seeds (42, 123, 456), five mask repeats, 200 bootstrap replicates, both split settings, five models, three missingness mechanisms, and 5%, 10%, and 20% masking.

**Before starting:** in the next cell upload these four files: `run_public_health_fmg_audit.py`, `requirements.txt`, the CDC 2024 BRFSS ZIP, and the UCI Diabetes 130-US Hospitals ZIP. You can download the script and requirements beside this notebook in the repository. The notebook identifies the data archives by their internal files.

The full CPU run may take several hours. Keep the Colab runtime connected. If interrupted, rerun the audit cell with the same `/content/outputs_full_v3` folder; the script resumes completed work from its checkpoint. Download the output ZIP before closing the runtime.

If you see code that automatically deletes `/content/outputs_full` and starts again after an error, stop that cell. This notebook does not retry or delete results automatically.

```python
from google.colab import files
from pathlib import Path
import zipfile, shutil

uploaded = files.upload()
root = Path('/content')
script_path = None
requirements_path = None
brfss_path = None
diabetes_path = None

for name in uploaded:
    src = root / name
    low = name.lower()
    if low.endswith('.py') and 'run_public_health_fmg_audit' in low:
        script_path = root / 'run_public_health_fmg_audit.py'
        if src != script_path: shutil.copy2(src, script_path)
    elif low.endswith('.txt') and 'requirement' in low:
        requirements_path = root / 'requirements.txt'
        if src != requirements_path: shutil.copy2(src, requirements_path)
    elif low.endswith('.zip'):
        with zipfile.ZipFile(src) as zf:
            members = [m.lower().split('/')[-1] for m in zf.namelist()]
        if any(m.endswith('.xpt') for m in members):
            brfss_path = root / 'brfss_2024.zip'
            if src != brfss_path: shutil.copy2(src, brfss_path)
        elif any('diabetic_data.csv' in m for m in members):
            diabetes_path = root / 'diabetes_130_hospitals.zip'
            if src != diabetes_path: shutil.copy2(src, diabetes_path)

missing = [label for label, value in [('audit script', script_path), ('requirements.txt', requirements_path), ('CDC BRFSS XPT ZIP', brfss_path), ('UCI diabetes ZIP', diabetes_path)] if value is None]
if missing:
    raise ValueError('Missing upload(s): ' + ', '.join(missing) + '. Upload the four requested files and rerun this cell.')
print('Uploads ready:')
print(' Script:', script_path)
print(' Requirements:', requirements_path)
print(' CDC data:', brfss_path)
print(' UCI data:', diabetes_path)
```

```python
import subprocess, sys
from pathlib import Path

# Redefine the file paths in case the runtime was restarted and in-memory variables were lost
script_path = Path('/content/run_public_health_fmg_audit.py')
brfss_path = Path('/content/brfss_2024.zip')
diabetes_path = Path('/content/diabetes_130_hospitals.zip')

command = [
    sys.executable, str(script_path),
    '--brfss', str(brfss_path),
    '--diabetes-zip', str(diabetes_path),
    '--seeds', '42', '123', '456',
    '--mask-repeats', '5',
    '--bootstrap-reps', '200',
    '--output-dir', '/content/outputs_full_v3',
]
print('Starting full run. Rerun this cell after an interruption to resume from checkpoint.')
print('Command:', ' '.join(command))
subprocess.run(command, check=True)
```

```python
from pathlib import Path
import json, pandas as pd

out = Path('/content/outputs_full_v3')

if (out / 'metrics_per_mask.csv').exists():
    metrics = pd.read_csv(out / 'metrics_per_mask.csv')
    print('Progress Metrics:')
    print('  Rows processed:', len(metrics))
    print('  Datasets:', sorted(metrics['dataset'].unique()))
    print('  Splits:', sorted(metrics['split'].unique()))
    print('  Models:', sorted(metrics['model'].unique()))
    print('  Mechanisms:', sorted(metrics['mechanism'].unique()))
    print('  Mask rates:', sorted(metrics['nominal_mask_rate'].unique()))
else:
    print('No metrics file found yet.')

if (out / 'run_manifest.json').exists():
    manifest = json.loads((out / 'run_manifest.json').read_text())
    print('\nRun Status: Completed')
    print('  Runtime (seconds):', manifest.get('runtime_seconds'))
    print('  Hardware:', manifest.get('hardware'))
else:
    print('\nRun Status: Still running / incomplete (no manifest generated yet)')

if (out / 'checkpoint.json').exists():
    checkpoint_data = json.loads((out / 'checkpoint.json').read_text())
    print('  Checkpoint completed units:', checkpoint_data.get('completed_units'))
```
